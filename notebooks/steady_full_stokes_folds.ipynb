# Steady full-Stokes generation of recumbent folds

This notebook tests whether a steady velocity field can generate an overturned isochrone directly from an ordered surface-deposition map. No finite-amplitude layer perturbation is prescribed. Each plotted isochrone is the set of particles with the same time since deposition, parameterized by its surface origin $x_0$. The direct fold test is $\partial x/\partial x_0<0$.

This is a mechanism calculation rather than an Allan Hills reconstruction. Length is in m, time in yr, velocity in m yr$^{-1}$, stress in Pa, and the vertical axis is elevation above the mean bed.

## Equations, assumptions, and boundary conditions

We solve the two-dimensional incompressible, linear-viscous full-Stokes equations

$$-\nabla\cdot[2\eta\,\mathrm{sym}(\nabla\mathbf{u})-p\mathbf{I}]=\rho\mathbf{g},\qquad \nabla\cdot\mathbf{u}=0,$$

using continuous quadratic velocity and continuous linear pressure. The domain is periodic over 800 m and has a flat surface at 160 m. A smooth raised-cosine obstacle forms the bed. Ice has $\eta=2.5\times10^8$ Pa yr and $\rho=917$ kg m$^{-3}$, and gravity acts at a mean surface slope of 0.03 rad. The surface is traction free.

The basal velocity is prescribed as $\mathbf{u}_b=U_b[1,b_x]^\mathsf{T}$ with $U_b=0.20$ m yr$^{-1}$. This is a sliding end member: it enforces no penetration of the fixed bed while allowing form drag to control the interior deformation. The flat surface is held steady by the diagnosed accumulation rate $\dot a=-w|_{z=160\,\mathrm{m}}$. Trajectories are used only where $\dot a>0$ and $u>0$.

The first obstacle isolates cumulative extrusion while horizontal speed remains comparable to its value at deposition. The second contains a much slower trajectory upstream of a steep obstacle and tests the residence-time weighting in the order budget. The observational prediction of both models is a vertical tangent followed by a backward-running isochrone limb.

## Interpretation

The colored curves are calculations, not observations. Their surface-deposition coordinates begin in increasing order. A negative value of $\partial x/\partial x_0$ therefore identifies a recumbent limb generated by the steady velocity field rather than the amplification of an inherited wrinkle.

The two calculations separate mechanisms within the same order theorem. In the persistent-extrusion case, reverse shear exceeds the initial order contribution without a reduction in horizontal speed below its deposition value. In the slow-flow case, the horizontal speed decreases substantially and the $u^{-2}$ travel-time weighting amplifies the reverse-shear contribution. The calculations establish that these boundary conditions permit steady, boundary-generated folding. They do not show that either idealized geometry occurred at Allan Hills.
